# Correctness

Auxiliary functions (part 1).

```python
# computing the gcd of vector components

def gcd_vec(vals):
    """
    vals can be a Python list, Sage vector, or free-module element.
    Returns gcd of entries as a nonnegative ZZ integer.
    """
    g = ZZ(0)
    for x in vals:
        g = gcd(g, ZZ(x))
    return abs(g)


# making a vector v as primitive by dividing the components with the gcd of the components.
def make_primitive(v):
    """
    Return primitive vector by dividing all coordinates by gcd.
    Works for free-module elements and vectors.
    """
    g = gcd_vec(v)
    if g == 0:
        # v is the zero vector
        return vector(ZZ, list(v))
    return vector(ZZ, [ZZ(x)//g for x in v])


# computing the Hamming weight
def hamming_weight(v):
    return sum(1 for x in v if ZZ(x) != 0)

# check whether a vector has -1/+1 components
def has_pm1(v):
    return any(abs(ZZ(x)) == 1 for x in v)

# check whether the entries of a vector are-1/0/+1
def in_pm1(v):
    for x in v:
        if ZZ(x) not in [-1, 0, 1]:
            return False
    return True
```

Auxiliary functions (part 2).

```python
# getting number of disjoint supports
def get_b(List_of_T):
    return len(List_of_T)

def support_sizes(Lists_of_T):
    return [len(T) for T in Lists_of_T]

def free_indices(List_of_T,n):
    T_new = []
    for sublist in List_of_T:
        T_new += sublist
    T_new.sort()
    Free = [i for i in range(n) if i not in T_new]
    return Free

def k_c_det(List_of_T,n):
    b = get_b(List_of_T)
    return n - b + 1

# list of hamming weight
def hamming_weight_list(List_of_v):
    return [hamming_weight(v) for v in List_of_v]

# finding the witness_size
def witness_sizes(Lambdas):
    return hamming_weight_list(Lambdas)
```

## Lower bound from disjoint supports

To analyze the theoretical lower bound of the success probability of the correctness if we construct $\mathbf{g}$ using the above algorithm, we consider a fixed $\mathbf{G} \in \mathbb{Z}^{n \times k}$. Observe that for $S \subseteq [1,n]$ and planted disjoint certified supports $T_1,\dots,T_b$, we have $\Pr\left[\mathbf{e}_1^\top \in \mathrm{Row}_\mathbb{Z}(\mathbf{G}_S)\right] \geq \Pr\left[\bigcup_{i=1}^b (T_i \subseteq S)\right]$. For a fixed planted certified support $T_i$ of size $|T_i| = t_i$ and fixed $S \subseteq [1,n]$ of size $|S| = s$, the probability that $T_i \subseteq S$ is $\Pr[T_i \subseteq S] = \frac{\binom{n-t_i}{s-t_i}}{\binom{n}{s}}$. Using the inclusion-exclusion principle, we have
\begin{align}
    \Pr\left[\bigcup_{i=1}^b (T_i \subseteq S)\right] &= \sum_{\emptyset \neq J \subseteq [1,b]} (-1)^{|J| + 1} \cdot \frac{\binom{n-\sum_{i \in J}t_i}{s-\sum_{i\in J} t_i}} {\binom{n}{s}} \nonumber \\ 
    &= \frac{1}{\binom{n}{s}} \sum_{\emptyset \neq J \subseteq [1,b]} (-1)^{|J| + 1} \cdot \binom{n-\sum_{i \in J}t_i}{s-\sum_{i\in J} t_i}. 
\end{align}
For $n \in \mathbb{Z}^+$ and $\mathcal{T} = \{T_1,\dots,T_b\}$ of disjoint supports where $T_i \subseteq [1,n]$, we can compute the value of $\Pr_{LB}(s,\mathcal{T})$ for $s \in [1,n]$ where
\begin{align}
    \Pr_{LB}(s,\mathcal{T}) \stackrel{\text{def}}{=} \frac{1}{\binom{n}{s}}\cdot\sum_{\emptyset \neq J \subseteq [1,b]} (-1)^{|J| + 1} \cdot \binom{n-\sum_{i \in J}t_i}{s-\sum_{i\in J} t_i},
\end{align}
where $t_i = |T_i|$ for $i \in [1,b]$.

```python
def PLB(n, s, sizes):
    """
    Exact Pr_LB(s,{T1,...,Tb}) using inclusion-exclusion.
    sizes = [t1,...,tb], disjoint supports.
    """
    b = len(sizes)
    denom = binomial(n, s)
    total = ZZ(0)

    # Iterate over nonempty subsets J via bitmasks
    for mask in range(1, 1 << b):
        u = 0
        bits = 0
        for i in range(b):
            if (mask >> i) & 1:
                u += sizes[i]
                bits += 1
        if u > s:
            continue

        term = binomial(n - u, s - u)
        if bits % 2 == 1:
            total += term
        else:
            total -= term

    return float(total / denom)
```

Constructing the curve for $\Pr_{LB}(s,\mathcal{T})$ for a given $\mathcal{T}$ and $1\leq s \leq n$. 

```python
def PLB_curve(n, sizes, s_min=1, s_max=None):
    """
    Return the PLB curve as a list of pairs [(s, PLB(n,s,sizes)), ...].

    Parameters
    ----------
    n : int
        Total number of indices.
    sizes : list of int
        sizes = [t1,...,tb] with disjoint supports.
    s_min : int
        Minimum s to include (default 1).
    s_max : int or None
        Maximum s to include, inclusive (default n).

    Returns
    -------
    list of tuples
        [(s, QQ probability), ...]
    """
    n = int(n)
    sizes = list(sizes)

    if s_max is None:
        s_max = n

    s_min = int(s_min)
    s_max = int(s_max)

    # Clamp the range safely
    if s_min < 0:
        s_min = 0
    if s_max > n:
        s_max = n

    curve = []
    for s in range(s_min, s_max + 1):
        curve.append((s, PLB(n, s, sizes)))
    return curve
```

## Lower bound from witness supports

Suppose $\mathcal{T}_i = \{T_1,\dots,T_b\}$ is a set of certified planted supports. We have $\bm{\lambda}_{T_i}^\top (\mathbf{g})_{T_i} = 1$. Let
\begin{align}
    W_i &\stackrel{\text{def}}{=} \mathrm{supp}(\bm{\lambda}_{(T_i)})\\
    w_i &\stackrel{\text{def}}{=} |W_i| = wt_H(\bm{\lambda}_{T_i})
\end{align}
If $W_i \subseteq S$, then we can embed $\bm{\lambda}_{T_i}$ into a vector $\bm{\lambda}_S\in \mathbb{Z}^{|S|}$ by placing non-zero entries on the coordinates of $W_i \subseteq S$ and zeros elsewhere. We have $\bm{\lambda}_S \in \ker_\mathbb{Z}(\mathbf{H}_{(S)})$. Since $\bm{\lambda}[j] = 0$ for $j \notin W_i$, we have $\bm{\lambda}_S^\top (\mathbf{g})_{S} = \bm{\lambda}_{T_i}^\top (\mathbf{g})_{T_i} = 1$. This makes $W_i \subseteq S \Rightarrow \mathbf{e}_1^\top \in \mathrm{Row}_\mathbb{Z}(\mathbf{G}_S)$.

Since $T_i$ for $i \in [1,b]$ are disjoint and $W_i \subseteq T_i$, $W_i$ for $i \in [1,b]$ are also disjoint, we have $\Pr[W_i \subseteq S] \geq \Pr[T_i \subseteq S]$ for random $S\subseteq [1,n]$. For $n \in \mathbb{Z}^+$, $s \in [1,n]$, and a collection of disjoint sets $\mathcal{W} = \{W_1,\dots,W_b\}$ where $W_i = \mathrm{supp}(\bm{\lambda}_{T_i})$, we define
\begin{align}
    \Pr_{witness}(s,\mathcal{W}) \stackrel{\text{def}}{=} \frac{1}{\binom{n}{s}}\cdot\sum_{\emptyset \neq J \subseteq [1,b]} (-1)^{|J| + 1} \cdot \binom{n-\sum_{i \in J}w_i}{s-\sum_{i\in J} w_i},
\end{align}
where $w_i = |W_i|$ for $i \in [1,b]$. We have $\Pr_{witness}(s,\mathcal{W}) \geq \Pr_{LB}(s,\mathcal{T})$.

```python
def PWitness(n, s, witness_sizes):
    """
    Exact Pr_witness(s,{W1,...,Wb}) using inclusion-exclusion.
    witness_sizes = [w1,...,wb], disjoint supports.
    """
    b = len(witness_sizes)
    denom = binomial(n, s)
    total = ZZ(0)

    # Iterate over nonempty subsets J via bitmasks
    for mask in range(1, 1 << b):
        u = 0
        bits = 0
        for i in range(b):
            if (mask >> i) & 1:
                u += witness_sizes[i]
                bits += 1
        if u > s:
            continue

        term = binomial(n - u, s - u)
        if bits % 2 == 1:
            total += term
        else:
            total -= term

    return float(total / denom)
```

Constructing the curve for $\Pr_{witness}(s,\mathcal{W})$ for a given $\mathcal{W}$ and $1\leq s \leq n$. 

```python
def PWitness_curve(n, witness_sizes, s_min=1, s_max=None):
    """
    Return the PWitness curve as a list of pairs [(s, PWitness(n,s,witness_sizes)), ...].

    Parameters
    ----------
    n : int
        Total number of indices.
    witness_sizes : list of int
        witness_sizes = [w1,...,wb] associated with disjoint supports.
    s_min : int
        Minimum s to include (default 1).
    s_max : int or None
        Maximum s to include, inclusive (default n).

    Returns
    -------
    list of tuples
        [(s, QQ probability), ...]
    """
    n = int(n)
    witness_sizes = list(witness_sizes)

    if s_max is None:
        s_max = n

    s_min = int(s_min)
    s_max = int(s_max)

    # Clamp the range safely
    if s_min < 0:
        s_min = 0
    if s_max > n:
        s_max = n

    curve = []
    for s in range(s_min, s_max + 1):
        curve.append((s, PWitness(n, s, witness_sizes)))
    return curve
```

# Empirical privacy analysis: proportion of $U$ that completely leaks the secret

Suppose we consider a fixed $U \subseteq [1,n]$. Then, $U$ completely leaks the secret if $\mathbf{e}_1^\top \in \mathrm{Row}_\mathbb{Q}(\mathbf{G}_U)$, i.e., $\mathbf{e}_1^\top$ can be expressed as a $\mathbb{Q}$-linear combination of the rows of $\mathbf{G}_U$. We can prove the following lemma.

**Lemma**: Fix $U \subseteq [1,n]$ and the matrix $\mathbf{G}$. The following conditions are equivalent.
1. $\mathbf{e}_1^\top \in \mathrm{Row}_\mathbb{Q}(\mathbf{G}_U)$.
2. If $\mathbf{F}_{\mathbf{G}_U}$ is the canonical reduced row echelon form (RREF) of $\mathbf{G}_U$, then $\mathbf{F}_{\mathbf{G}_U}[1,\cdot] = \mathbf{e}_1^\top = (1,0,\dots,0) \in \mathbb{Q}^k$.

We define the function **rational_span(G)** that takes a matrix $\mathbf{G} \in \mathbb{Z}^{s \times k}$ as input and returns true if $\mathbf{e}_1^\top \in \mathrm{Row}_\mathbb{Q}(\mathbf{G})$.

Suppose $\mathbf{G} \in \mathbb{Z}^{s \times k}$ and $\mathbf{e}_1^\top \in \mathrm{Row}_\mathbb{Q}(\mathbf{G})$. This means $\mathbf{e}_1^\top$ can be obtained using elementary row operations (over $\mathbb{Q}$) on $\mathbf{G}$. If $\mathbf{F}_\mathbf{G} \in \mathbb{Q}^{s \times k}$ is the RREF of $\mathbf{G}$, then it is unique lower triangular matrix such that $\mathbf{F}_\mathbf{G} = \mathbf{E}_\mathbf{G} \mathbf{G}$ for some invertible matrix $\mathbf{E}_\mathbf{G} \in \mathbb{Q}^{s \times s}$. By the RREF pivot condition, if $\mathbf{e}_1^\top \in \mathrm{Row}_\mathbb{Q}(\mathbf{G})$, we must have $\mathbf{F}_\mathbf{G}[1,\cdot] = \mathbf{e}_1^\top$ (i.e., the first row of $\mathbf{F}_\mathbf{G}$ ($1$-based indexing) is $\mathbf{e}_1^\top$. In short: $\mathbf{e}_1^\top \in \mathrm{Row}_\mathbb{Q}(\mathbf{G}) \Leftrightarrow \mathbf{F}_\mathbf{G}[1,\cdot] = \mathbf{e}_1^\top$. For fast testing, we consider the following equivalence:
\begin{align*}
    \mathbf{e}_1^\top \in \mathrm{Row}_\mathbb{Q}(\mathbf{G}) \Leftrightarrow \mathrm{rank}_\mathbb{Q}(\mathbf{G}) = \mathrm{rank}_\mathbb{Q}\left(\begin{bmatrix}\mathbf{e}_1^\top\\\mathbf{G}\end{bmatrix}\right)
\end{align*}

We define the function **rational_span_S(G,S)** that takes a matrix $\mathbf{G} \in \mathbb{Z}^{n \times k}$ and $S \subseteq [0,n-1]$ ($0$-based-index) as input and returns true if $\mathbf{e}_1^\top \in \mathrm{Row}_\mathbb{Q}(\mathbf{G}_S)$.

We define $\Pr_{\text{complete leak}}(s,\mathbf{G})$ for a fixed matrix $\mathbf{G} \in \mathbb{Z}^{n \times k}$ and $s \in [1,n]$ as
\begin{align}
    \Pr_{\text{complete leak}}(s,\mathbf{G}) &\stackrel{\text{def}}{=} \frac{|\{S\subseteq [1,n]: \left(|S| = s\right) \land \left(\mathbf{e}_1^\top \in \mathrm{Row}_\mathbb{Q}(\mathbf{G}_S)\right)\}|}{|\{S\subseteq [1,n]: \left(|S| = s\right)\}|}\\
    &= \frac{1}{\binom{n}{s}}|\{S\subseteq [1,n]: \left(|S| = s\right) \land \left(\mathbf{e}_1^\top \in \mathrm{Row}_\mathbb{Q}(\mathbf{G}_S)\right)\}|
\end{align}

## Checking that $\mathrm{Row}_\mathbb{Z}(\mathbf{H'}_{(T)}) = \mathbb{Z}^{|T|}$ for every $T \subseteq [1,n]$ with $|T| \leq k_p$

We build a function that check whether $\mathrm{Row}_\mathbb{Z}(\mathbf{A}) = \mathbb{Z}^{|T|}$ for $\mathbf{A} \in \mathbb{Z}^{m \times |T|}$ where $|T| \leq m$. We use SNF technique to verify this condition. If $\mathbf{A} \in \mathbb{Z}^{m \times |T|}$ with $\mathrm{rank}_\mathbb{Q}(\mathbf{A}) = r$, then according to SNF of $\mathbf{A}$, there are unimodular matrices $\mathbf{U} \in GL_m(\mathbb{Z}), \mathbf{V}\in GL_{|T|}(\mathbb{Z})$, and a diagonal matrix $\mathbf{D} \in \mathbb{Z}^{m \times |T|}$ with $\mathbf{D} = \mathrm{diag}(d_1,\dots,d_r,0,\dots,0)$ where $d_i > 0$ and $d_i \mid d_{i+1}$ for $i \in [1,r-1]$, such that $\mathbf{U}\mathbf{A}\mathbf{V}=\mathbf{D}$. We have $\mathbf{A} = \mathbf{U}^{-1}\mathbf{D}\mathbf{V}^{-1}$ and we get $\mathrm{Row}_\mathbb{Z}(\mathbf{A}) = \mathrm{Row}_\mathbb{Z}(\mathbf{D})$. Thus, $\mathrm{Row}_\mathbb{Z}(\mathbf{A}) = \mathrm{Row}_\mathbb{Z}(\mathbf{D}) = \mathbb{Z}^{|T|}$ if and only if $\mathrm{rank}_\mathbb{Q}(\mathbf{A}) = |T|$ and $d_1 = \cdots = d_{|T|} = 1$.

```python
def row_lattice_is_full_Z(A):
    """
    A: Sage integer matrix (m x t) over ZZ. Returns True iff Row_Z(A) = Z^t.
    """
    m, t = A.nrows(), A.ncols()
    if t == 0:
        return True
    if A.rank() < t:   # rank over QQ
        return False
    D, U, V = A.smith_form()   # D is diagonal Smith form
    return all(D[i, i] == 1 for i in range(t))
```

```python
from itertools import combinations

def check_row_span_property(H, gamma, eps, return_counterexample=False):
    """
    Checks: for all T ⊆ [n] with 1 ≤ |T| < 2*gamma*(1-eps)*n, Row_Z(H_(T)) = Z^{|T|}.
    H: Sage matrix (m x n) over GF(2), ZZ, etc. with integer entries.
    gamma: real/rational in [0,1]
    eps: real/rational in (0,1/2)
    """
    m, n = int(H.nrows()), int(H.ncols()) 
    K1 = int(floor(gamma * n))
    K2 = int(floor(2*gamma*(1-eps)*n))-1
    K = max(K1,K2)

    if K <= 0:
        return (True, {"checked_up_to": 0}) if return_counterexample else True

    # Work over ZZ
    HZ = Matrix(ZZ, H)

    for t in range(1, K + 1):
        for T in combinations(range(n), t):      # 0-based column indices
            A = HZ[:, list(T)]                  # m x t submatrix H_(T)
            if not row_lattice_is_full_Z(A):
                if return_counterexample:
                    return False, {
                        "reason": "row_span_not_full",
                        "bad_T_0_based": list(T),
                        "size": t
                    }
                return False

    return (True, {"checked_up_to": K}) if return_counterexample else True
```

## Computing $\alpha_{k_p} = \min_{\{U:|U| = k_p\}}\sigma_{\min}(\mathbf{\hat{G}}_U)$

```python
from itertools import combinations
import numpy as np

def alpha_kp_exact(Ghat, k_p, verbose=False, report_every=None):
    r"""
    Exact enumeration of alpha_{k_p} = min_{|U|=k_p} sigma_min(Ghat_U).

    Optimised for n ~ 40, k_p in {5, 6}: uses NumPy directly on a
    pre-converted dense array to avoid per-iteration Sage overhead.

    INPUT:
    - ``Ghat``         -- integer matrix, n x (k-1)
    - ``k_p``          -- subset size
    - ``verbose``      -- if True, return (alpha, U_min)
    - ``report_every`` -- if int, print progress every this many subsets

    OUTPUT:
    - alpha_{k_p}, or (alpha_{k_p}, U_min) if verbose=True.
    """
    n = Ghat.nrows()
    assert 1 <= k_p <= n

    # Convert once to a NumPy float64 array; per-subset slicing is then
    # just NumPy indexing, and svd() goes straight to LAPACK.
    A = np.array(Ghat, dtype=np.float64)

    alpha = np.inf
    U_min = None
    total = int(binomial(n, k_p))
    print(f"Enumerating {total} subsets of size {k_p}...")

    for i, U in enumerate(combinations(range(n), k_p)):
        sub = A[list(U), :]
        # compute_uv=False returns just the singular values (descending)
        s_min = np.linalg.svd(sub, compute_uv=False)[-1]
        if s_min < alpha:
            alpha = s_min
            U_min = U
        if report_every and (i + 1) % report_every == 0:
            print(f"  {i+1}/{total} done, current min = {alpha:.6g}")

    return (float(alpha), U_min) if verbose else float(alpha)
```

## Computing $\beta_{k_p} = \max_{\{U:|U| = k_p\}}\sigma_{\max}(\mathbf{\hat{G}}_U) = \max_{\{U:|U| = k_p\}}\|\mathbf{\hat{G}}_U\|_2$ 

```python
from itertools import combinations
import numpy as np

def beta_kp_exact(Ghat, k_p, verbose=False, report_every=None):
    r"""
    Exact enumeration of beta_{k_p} = max_{|U|=k_p} sigma_max(Ghat_U).
    INPUT:
    - ``Ghat``         -- integer matrix, n x (k-1)
    - ``k_p``          -- subset size
    - ``verbose``      -- if True, return (beta, U_max)
    - ``report_every`` -- if int, print progress every this many subsets
    OUTPUT:
    - beta_{k_p}, or (beta_{k_p}, U_max) if verbose=True.
    """
    n = Ghat.nrows()
    assert 1 <= k_p <= n
    A = np.array(Ghat, dtype=np.float64)
    beta = -np.inf
    U_max = None
    total = int(binomial(n, k_p))
    print(f"Enumerating {total} subsets of size {k_p}...")
    for i, U in enumerate(combinations(range(n), k_p)):
        sub = A[list(U), :]
        s_max = np.linalg.svd(sub, compute_uv=False)[0]   # largest, index 0
        if s_max > beta:
            beta = s_max
            U_max = U
        if report_every and (i + 1) % report_every == 0:
            print(f"  {i+1}/{total} done, current max = {beta:.6g}")
    return (float(beta), U_max) if verbose else float(beta)
```

## Computing $\gamma_{k_p} = \min_{\{U:|U| \leq k_p\}}\sigma_{\min}(\mathbf{M}_U)$

$\mathbf{M}_U = \mathbf{\hat{G}}_U[\cdot,J_U] \in GL_{|U|}(\mathbb{Z})$ is a unimodular $|U| \times |U| $ minor of $\mathbf{\hat{G}}_U$ obtained using the following algorithm:
1. Input a matrix $\mathbf{\hat{G}}_U \in \mathbb{Z}^{r \times (k-1)}$, where $r := |U|$, satisfying the unit-valued unique-neighbor property from expander code: $\forall \emptyset \neq S \subseteq[r], \exists j \in [k-1]: |\{i \in S: \mathbf{\hat{G}}_U[i,j] \neq 0\}| = 1$, and the unique non-zero entry belongs to $\{-1,1\}$.
2. Output: An ordered tuple $J_U = (j_1,\dots,j_r)$ such that $\mathbf{M}_U:= \mathbf{\hat{G}}_U[\cdot,J_U] \in GL_r(\mathbb{Z})$.
3. $S := [r]$, $J_U := ()$, $I_U := ()$
4. for $t = 1$ to $r$:
   1. for $j \in [k-1] \setminus \{j_1,\dots,j_{t-1}\}$: $\mathrm{supp}_S(j):= \{i \in S: \mathbf{\hat{G}}_U[i,j] \neq 0\}$.
   2. $C := \{j \in [k-1]\setminus \{j_1,\dots,j_{t-1}\}: |\mathrm{supp}_S(j)| =1, \mathbf{\hat{G}}_U[i(j),j] \in \{-1,1\}\}$ where $i(j)$ is the unique row i $\mathrm{supp}_S(j)$.
   3. $j_t = \min C$
   4. Let $i_t$ be the unique row in $\mathrm{supp}_S(j_t)$
   5. $J_U := (j_1,\dots,j_{t-1},j_t)$, $I_U:= (i_1,\dots,i_{t-1},i_t)$
   6. $S:= S \setminus \{i_t\}$
5. $\mathbf{M}_U:= \mathbf{\hat{G}}_U[\cdot,J_U]$
6. return $(\mathbf{M}_U,J_U,I_U)$

```python
def construct_MU(GhatU):
    r"""Algorithm:Construct-MU — peeling-based unimodular minor.
    Returns (M_U, J_U, I_U); raises AssertionError if the unit unique-neighbor
    hypothesis fails (indicates a violated hypothesis or wrong matrix orientation).
    """
    GhatU = GhatU.change_ring(ZZ)          # force exact integer semantics
    r = GhatU.nrows()
    m = GhatU.ncols()
    S = set(range(r))
    J_U, I_U, J_set = [], [], set()
    for t in range(r):
        C = []
        for j in range(m):
            if j in J_set:
                continue
            supp = [i for i in S if GhatU[i, j] != 0]
            if len(supp) == 1 and GhatU[supp[0], j] in (1, -1):
                C.append((j, supp[0]))
        if not C:
            raise AssertionError(
                f"no unit unique-neighbor column at step {t}; "
                f"hypothesis violated or wrong orientation. "
                f"S={sorted(S)}, J_U={J_U}")
        j_t, i_t = min(C, key=lambda p: p[0])
        J_U.append(j_t); J_set.add(j_t)
        I_U.append(i_t)
        S.discard(i_t)
    M_U = GhatU.matrix_from_columns(J_U)
    return M_U, tuple(J_U), tuple(I_U)
```

Computing $\gamma_{k_p}$.

```python
from itertools import combinations
import numpy as np

def gamma_kp(Ghat, k_p, verbose=False, report_every=None, track_W=False):
    r"""
    gamma_{k_p} = min over 1 <= |U| <= k_p of sigma_min(M_U),
    where M_U is the deterministic minor from Algorithm:Construct-MU.

    INPUT
    - Ghat : n x (k-1) integer matrix. Rows index parties [n]; a subset U
             selects rows, so Ghat_U := Ghat[U, :] is |U| x (k-1).
    - k_p  : max subset size.
    - track_W : if True, also return max ||W||_2 over all U (the strictly-upper
                triangular part of M_U^triangle), for the analytic gamma floor.

    OUTPUT
    - gamma, arg_U            (and max_W if track_W)
    """
    Ghat = Ghat.change_ring(ZZ)
    n = Ghat.nrows()
    assert 1 <= k_p <= n
    gamma, arg = np.inf, None
    max_W, arg_W = 0.0, None

    total = sum(int(binomial(n, s)) for s in range(1, k_p + 1))
    print(f"Enumerating {total} subsets (sizes 1..{k_p})...")
    seen = 0
    for size in range(1, k_p + 1):           # ALL sizes: sigma_min not monotone in |U|
        for U in combinations(range(n), size):
            GhatU = Ghat.matrix_from_rows(list(U))          # |U| x (k-1)
            M_U, J_U, I_U = construct_MU(GhatU)             # raises if hypothesis fails

            # stable sigma_min(M_U) = 1 / sigma_max(M_U^{-1}), M_U^{-1} exact integer
            Minv = M_U.inverse()
            s_min = 1.0 / np.linalg.svd(np.array(Minv, dtype=np.float64),
                                        compute_uv=False)[0]
            if s_min < gamma:
                gamma, arg = s_min, U

            if track_W:
                # M_U^triangle = GhatU[I_U, J_U]; W = D^{-1}(tri - D), D=diag(+-1)
                Tri = GhatU.matrix_from_rows_and_columns(list(I_U), list(J_U))
                D = diagonal_matrix(ZZ, [Tri[a, a] for a in range(size)])
                W = D.inverse() * (Tri - D)                 # strictly upper, {0,+-1}
                w2 = np.linalg.svd(np.array(W, dtype=np.float64),
                                   compute_uv=False)[0] if size > 1 else 0.0
                if w2 > max_W:
                    max_W, arg_W = w2, U

            seen += 1
            if report_every and seen % report_every == 0:
                print(f"  {seen}/{total}, gamma so far = {gamma:.6g}")

    if track_W:
        return float(gamma), arg, float(max_W), arg_W
    return (float(gamma), arg) if verbose else float(gamma)
```

Faster approach for computing $\gamma_{k_p}$.

```python
from itertools import combinations
import numpy as np

def gamma_kp_fast(Ghat, k_p, report_every=None, track_W=False):
    r"""
    gamma_{k_p} = min_{1<=|U|<=k_p} sigma_min(M_U), M_U from Algorithm:Construct-MU.
    Fully vectorized: Ghat converted once to int8 NumPy; peeling uses count_nonzero;
    sigma_min taken directly on the triangular +-1 block M_U^triangle (accurate for
    small kp). Exact cross-check the caller runs afterward on the returned arg_U.
    """
    A = np.array(Ghat.change_ring(ZZ), dtype=np.int8)   # n x m, entries in {0,+-1}
    n, m = A.shape
    assert 1 <= k_p <= n

    gamma, arg = np.inf, None
    max_W, arg_W = 0.0, None
    total = sum(int(binomial(n, s)) for s in range(1, k_p + 1))
    print(f"Enumerating {total} subsets (sizes 1..{k_p})...")
    seen = 0

    for size in range(1, k_p + 1):
        for U in combinations(range(n), size):
            sub_full = A[list(U), :]                    # size x m  (all cols)
            # --- peeling on the support, vectorized ---
            active = np.ones(size, dtype=bool)
            used = np.zeros(m, dtype=bool)
            I_U = np.empty(size, dtype=np.intp)
            J_U = np.empty(size, dtype=np.intp)
            ok = True
            for t in range(size):
                cnt = np.count_nonzero(sub_full[active, :], axis=0)  # per-col support
                cand = np.where((cnt == 1) & (~used))[0]
                picked = False
                rows_act = np.where(active)[0]
                for j in cand:                          # leftmost valid = min index
                    i_loc = np.nonzero(sub_full[active, j])[0][0]
                    i_glob = rows_act[i_loc]
                    if sub_full[i_glob, j] in (1, -1):
                        I_U[t], J_U[t] = i_glob, j
                        used[j] = True
                        active[i_glob] = False
                        picked = True
                        break
                if not picked:
                    ok = False
                    break
            if not ok:
                raise AssertionError(f"no unit unique-neighbor col, U={U}")

            # --- sigma_min via the triangular +-1 block directly (float64, accurate) ---
            Tri = sub_full[np.ix_(I_U, J_U)].astype(np.float64)   # size x size, triangular
            s_min = np.linalg.svd(Tri, compute_uv=False)[-1]
            if s_min < gamma:
                gamma, arg = s_min, U

            if track_W and size > 1:
                D = np.sign(np.diag(Tri))               # +-1 diagonal
                W = (Tri - np.diag(np.diag(Tri))) * D[:, None]  # strictly upper, {0,+-1}
                w2 = np.linalg.svd(W, compute_uv=False)[0]
                if w2 > max_W:
                    max_W, arg_W = w2, U

            seen += 1
            if report_every and seen % report_every == 0:
                print(f"  {seen}/{total}, gamma so far = {gamma:.6g}")

    if track_W:
        return float(gamma), arg, float(max_W), arg_W
    return float(gamma), arg
```

Faster approach for computing $\gamma_{k_p}$ using bitmask supports, pivot memoization, triangular caching, batched spectral evaluation.

```python
from itertools import combinations
import numpy as np


def _popcount(x):
    try:
        return x.bit_count()
    except AttributeError:
        return bin(int(x)).count("1")


def gamma_kp_bitset(Ghat, k_p, report_every=None, track_W=False,
                    batch_size=4096, keep_near=0, full=False):
    r"""
    gamma_{k_p} = min_{1 <= |U| <= k_p} sigma_min(M_U), M_U from peeling.
    Bitmask peeling + pivot memoization + triangular caching + batched SVD.
    Returns (gamma, arg_U); with track_W (gamma, arg_U, nu, arg_U_W);
    with full=True a diagnostics dict.
    """
    ONE = int(1)                                   # preparser-proof Python int

    A = np.asarray(Ghat.change_ring(ZZ), dtype=np.int8)
    assert A.ndim == 2, "Ghat must be a matrix."
    n, m = A.shape
    assert 1 <= k_p <= n, "require 1 <= k_p <= n."
    assert np.all(np.isin(A, (-1, 0, 1))), "entries must be in {0, +/-1}."

    # --- column support bitmasks (Python ints) ---
    column_masks = []
    for j in range(m):
        mask = 0
        for i in np.flatnonzero(A[:, j]):
            mask |= (ONE << int(i))
        column_masks.append(int(mask))

    # --- memoized deterministic peeling pivot ---
    pivot_cache = {}

    def leftmost_pivot(active_mask):
        cached = pivot_cache.get(active_mask)
        if cached is not None:
            return cached
        for j, col_mask in enumerate(column_masks):
            support = col_mask & active_mask
            if _popcount(support) == 1:
                i = int(support).bit_length() - 1
                pivot_cache[active_mask] = (j, i)
                return j, i
        active_rows = tuple(i for i in range(n) if (active_mask >> i) & 1)
        raise AssertionError(
            f"no unit unique-neighbor column for active rows {active_rows}; "
            f"hypothesis violated or wrong orientation.")

    def peel(U):
        active_mask = 0
        for i in U:
            active_mask |= (ONE << int(i))
        r = len(U)
        I = np.empty(r, dtype=np.intp)
        J = np.empty(r, dtype=np.intp)
        for t in range(r):
            j_t, i_t = leftmost_pivot(active_mask)
            i_t = int(i_t)
            I[t], J[t] = i_t, j_t
            active_mask &= ~(ONE << i_t)
        Tri = A[np.ix_(I, J)].astype(np.float64, copy=False)
        return I, J, Tri

    # --- state ---
    gamma, arg_U, arg_I, arg_J, arg_Tri = np.inf, None, None, None, None
    min_sizes = set()
    nu, arg_U_W = 0.0, None
    sigma_cache = {}
    near_candidates = []
    pending = {}

    def register(sigma, U, I, J, Tri, sizes):
        nonlocal gamma, arg_U, arg_I, arg_J, arg_Tri, min_sizes, near_candidates
        sigma = float(sigma)
        if sigma < gamma - 1e-15:
            gamma = sigma
            arg_U = tuple(U)
            arg_I = tuple(int(x) for x in I)
            arg_J = tuple(int(x) for x in J)
            arg_Tri = np.array(Tri, copy=True)
            min_sizes = set(sizes)
        elif abs(sigma - gamma) <= 1e-15:
            min_sizes |= set(sizes)
        if keep_near > 0:
            near_candidates.append({"sigma_min": sigma, "U": tuple(U),
                                    "I": tuple(int(x) for x in I),
                                    "J": tuple(int(x) for x in J),
                                    "Tri": np.array(Tri, copy=True)})
            near_candidates.sort(key=lambda x: x["sigma_min"])
            del near_candidates[keep_near:]

    def register_W(w2, U):
        nonlocal nu, arg_U_W
        if float(w2) > nu:
            nu, arg_U_W = float(w2), tuple(U)

    def flush():
        nonlocal pending
        if not pending:
            return
        keys = list(pending)
        mats = np.stack([pending[k]["Tri"] for k in keys], axis=0)
        sig = np.linalg.svd(mats, compute_uv=False)[:, -1]
        if track_W:
            r = mats.shape[1]
            W = mats.copy()
            idx = np.arange(r)
            W[:, idx, idx] = 0.0
            w2 = np.linalg.svd(W, compute_uv=False)[:, 0]
        for b, k in enumerate(keys):
            info = pending[k]
            U, I, J = info["rep"]
            sigma_cache[k] = float(sig[b])
            register(sig[b], U, I, J, info["Tri"], info["sizes"])
            if track_W:
                register_W(w2[b], U)
        pending = {}

    total = sum(int(binomial(n, s)) for s in range(1, k_p + 1))
    print(f"Enumerating {total} subsets of sizes 1,...,{k_p}.")
    seen = 0

    for size in range(1, k_p + 1):
        for U in combinations(range(n), size):
            I, J, Tri = peel(U)
            key = (int(size), Tri.astype(np.int8).tobytes())
            cached = sigma_cache.get(key)
            if cached is not None:
                register(cached, U, I, J, Tri, {int(size)})
            elif key in pending:
                pending[key]["sizes"].add(int(size))
            else:
                pending[key] = {"Tri": Tri, "sizes": {int(size)}, "rep": (U, I, J)}
                if len(pending) >= batch_size:
                    flush()
            seen += 1
            if report_every and seen % report_every == 0:
                flush()
                print(f"  {seen}/{total}; gamma so far = {gamma:.12g}; "
                      f"pivot states = {len(pivot_cache)}; "
                      f"distinct minors = {len(sigma_cache)}")
        flush()

    flush()

    if full:
        return {"gamma": float(gamma), "arg_U": arg_U, "arg_I": arg_I,
                "arg_J": arg_J, "arg_Tri": arg_Tri, "min_sizes": min_sizes,
                "nu": float(nu), "arg_U_W": arg_U_W,
                "near_candidates": near_candidates,
                "pivot_cache_size": len(pivot_cache),
                "sigma_cache_size": len(sigma_cache),
                "total_subsets": total}
    if track_W:
        return float(gamma), arg_U, float(nu), arg_U_W
    return float(gamma), arg_U
```

```python
from itertools import combinations
from collections import OrderedDict
import numpy as np


def _popcount(x):
    try:
        return x.bit_count()
    except AttributeError:
        return bin(int(x)).count("1")


def gamma_kp_bitset_LRUcapped(Ghat, k_p, report_every=None, track_W=False,
                    batch_size=4096, keep_near=0, full=False,
                    pivot_cache_max=20_000_000):
    r"""
    gamma_{k_p} = min_{1 <= |U| <= k_p} sigma_min(M_U), M_U from peeling.
    Bitmask peeling + LRU-capped pivot cache + triangular-minor caching
    + batched SVD.

    The pivot cache is LRU-capped at pivot_cache_max entries (~124 bytes each,
    so 20M -> ~2.5 GB) to retain the reuse speedup while bounding memory.
    Set pivot_cache_max=None for an unbounded cache (fast but may OOM at large n).

    Returns (gamma, arg_U); with track_W (gamma, arg_U, nu, arg_U_W);
    with full=True a diagnostics dict.
    """
    ONE = int(1)

    A = np.asarray(Ghat.change_ring(ZZ), dtype=np.int8)
    assert A.ndim == 2, "Ghat must be a matrix."
    n, m = A.shape
    assert 1 <= k_p <= n, "require 1 <= k_p <= n."
    assert np.all(np.isin(A, (-1, 0, 1))), "entries must be in {0, +/-1}."

    column_masks = []
    for j in range(m):
        mask = 0
        for i in np.flatnonzero(A[:, j]):
            mask |= (ONE << int(i))
        column_masks.append(int(mask))

    # --- LRU-capped pivot cache: keeps reuse speedup, bounds memory ---
    pivot_cache = OrderedDict()

    def leftmost_pivot(active_mask):
        cached = pivot_cache.get(active_mask)
        if cached is not None:
            pivot_cache.move_to_end(active_mask)      # mark recently used
            return cached
        for j, col_mask in enumerate(column_masks):
            support = col_mask & active_mask
            if _popcount(support) == 1:
                i = int(support).bit_length() - 1
                pivot_cache[active_mask] = (j, i)
                if pivot_cache_max is not None and len(pivot_cache) > pivot_cache_max:
                    pivot_cache.popitem(last=False)   # evict least-recently-used
                return j, i
        active_rows = tuple(i for i in range(n) if (active_mask >> i) & 1)
        raise AssertionError(
            f"no unit unique-neighbor column for active rows {active_rows}; "
            f"hypothesis violated or wrong orientation.")

    def peel(U):
        active_mask = 0
        for i in U:
            active_mask |= (ONE << int(i))
        r = len(U)
        I = np.empty(r, dtype=np.intp)
        J = np.empty(r, dtype=np.intp)
        for t in range(r):
            j_t, i_t = leftmost_pivot(active_mask)
            i_t = int(i_t)
            I[t], J[t] = i_t, j_t
            active_mask &= ~(ONE << i_t)
        Tri = A[np.ix_(I, J)].astype(np.float64, copy=False)
        return I, J, Tri

    gamma, arg_U, arg_I, arg_J, arg_Tri = np.inf, None, None, None, None
    min_sizes = set()
    nu, arg_U_W = 0.0, None
    sigma_cache = {}          # triangular-minor cache: small, keep unbounded
    near_candidates = []
    pending = {}

    def register(sigma, U, I, J, Tri, sizes):
        nonlocal gamma, arg_U, arg_I, arg_J, arg_Tri, min_sizes, near_candidates
        sigma = float(sigma)
        if sigma < gamma - 1e-15:
            gamma = sigma
            arg_U = tuple(U)
            arg_I = tuple(int(x) for x in I)
            arg_J = tuple(int(x) for x in J)
            arg_Tri = np.array(Tri, copy=True)
            min_sizes = set(sizes)
        elif abs(sigma - gamma) <= 1e-15:
            min_sizes |= set(sizes)
        if keep_near > 0:
            near_candidates.append({"sigma_min": sigma, "U": tuple(U),
                                    "I": tuple(int(x) for x in I),
                                    "J": tuple(int(x) for x in J),
                                    "Tri": np.array(Tri, copy=True)})
            near_candidates.sort(key=lambda x: x["sigma_min"])
            del near_candidates[keep_near:]

    def register_W(w2, U):
        nonlocal nu, arg_U_W
        if float(w2) > nu:
            nu, arg_U_W = float(w2), tuple(U)

    def flush():
        nonlocal pending
        if not pending:
            return
        keys = list(pending)
        mats = np.stack([pending[k]["Tri"] for k in keys], axis=0)
        sig = np.linalg.svd(mats, compute_uv=False)[:, -1]
        if track_W:
            r = mats.shape[1]
            W = mats.copy()
            idx = np.arange(r)
            W[:, idx, idx] = 0.0
            w2 = np.linalg.svd(W, compute_uv=False)[:, 0]
        for b, k in enumerate(keys):
            info = pending[k]
            U, I, J = info["rep"]
            sigma_cache[k] = float(sig[b])
            register(sig[b], U, I, J, info["Tri"], info["sizes"])
            if track_W:
                register_W(w2[b], U)
        pending = {}

    total = sum(int(binomial(n, s)) for s in range(1, k_p + 1))
    print(f"Enumerating {total} subsets of sizes 1,...,{k_p}.")
    seen = 0

    for size in range(1, k_p + 1):
        for U in combinations(range(n), size):
            I, J, Tri = peel(U)
            key = (int(size), Tri.astype(np.int8).tobytes())
            cached = sigma_cache.get(key)
            if cached is not None:
                register(cached, U, I, J, Tri, {int(size)})
            elif key in pending:
                pending[key]["sizes"].add(int(size))
            else:
                pending[key] = {"Tri": Tri, "sizes": {int(size)}, "rep": (U, I, J)}
                if len(pending) >= batch_size:
                    flush()
            seen += 1
            if report_every and seen % report_every == 0:
                flush()
                print(f"  {seen}/{total}; gamma so far = {gamma:.12g}; "
                      f"pivot entries = {len(pivot_cache)}; "
                      f"distinct minors = {len(sigma_cache)}")
        flush()

    flush()

    if full:
        return {"gamma": float(gamma), "arg_U": arg_U, "arg_I": arg_I,
                "arg_J": arg_J, "arg_Tri": arg_Tri, "min_sizes": min_sizes,
                "nu": float(nu), "arg_U_W": arg_U_W,
                "near_candidates": near_candidates,
                "pivot_cache_size": len(pivot_cache),
                "sigma_cache_size": len(sigma_cache),
                "total_subsets": total}
    if track_W:
        return float(gamma), arg_U, float(nu), arg_U_W
    return float(gamma), arg_U
```

## Computing $\delta_{k_p} = \max_{\{U:|U| = k_p\}}\sigma_{\max}(\mathbf{G}_U) = \max_{\{U:|U| = k_p\}}\|\mathbf{G}_U\|_2$ 

```python
from itertools import combinations
import numpy as np

def delta_kp_exact(G, k_p, verbose=False, report_every=None):
    r"""
    Exact enumeration of delta_{k_p} = max_{|U|=k_p} sigma_max(G_U).
    INPUT:
    - ``G``            -- integer matrix, n x k
    - ``k_p``          -- subset size
    - ``verbose``      -- if True, return (delta, U_max)
    - ``report_every`` -- if int, print progress every this many subsets
    OUTPUT:
    - delta_{k_p}, or (delta_{k_p}, U_max) if verbose=True.
    """
    n = G.nrows()
    assert 1 <= k_p <= n
    A = np.array(G, dtype=np.float64)
    delta = -np.inf
    U_max = None
    total = int(binomial(n, k_p))
    print(f"Enumerating {total} subsets of size {k_p}...")
    for i, U in enumerate(combinations(range(n), k_p)):
        sub = A[list(U), :]
        s_max = np.linalg.svd(sub, compute_uv=False)[0]   # largest, index 0
        if s_max > delta:
            delta = s_max
            U_max = U
        if report_every and (i + 1) % report_every == 0:
            print(f"  {i+1}/{total} done, current max = {delta:.6g}")
    return (float(delta), U_max) if verbose else float(delta)
```

## Computing the upper bound $\rho_{\max}$

Privacy of the scheme requires
\begin{align*}
    \sigma_\mathbf{R}^2 \geq \frac{\rho_{\max}}{\frac{1}{2\sigma_\mathrm{red}^2}-\frac{1}{\sigma_\mathbf{s}^2}}
\end{align*}
where $\rho_{\max} \leq \frac{k_p\beta_\mathbf{g}^2}{\alpha_{k_p}^2}$. We compute the upper bound for such $\rho_{\max}$ given $k_p$, $\beta_\mathbf{g}$, and $\alpha_{k_p}$ (obtained from previous calculation).

Note: we use the notation $\rho_{\max}$ in the write-up, while retaining $\kappa_{\max}$ in this script.

```python
def kappa_max_upper_bound(kp, beta_g, alpha_kp):
    """
    Upper bound for kappa_max:
        kappa_max <= k_p * beta_g^2 / alpha_{k_p}^2.
    """
    assert kp >= 1, "kp must be a positive integer."
    assert beta_g > 0, "beta_g must be positive."
    assert alpha_kp > 0, "alpha_kp must be positive."

    return (kp * beta_g**2) / (alpha_kp**2)
```
