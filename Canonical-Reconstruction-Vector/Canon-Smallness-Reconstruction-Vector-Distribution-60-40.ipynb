# Constructing Small Reconstruction Vector via Canonization Approach

We consider the approach to construct small reconstruction vector via canonization approach.

## Some Basic and Helper Algorithms

Helper functions.

```python
def linf_norm(v):
    """Infinity norm for a ZZ vector-like object."""
    if len(v) == 0:
        return ZZ(0)
    return max(abs(ZZ(x)) for x in v)

# computing the Hamming weight
def hamming_weight(v):
    return sum(1 for x in v if ZZ(x) != 0)
```

Sampling unique subsets.

```python
import random

# sampling N distinct random subsets of [0,n-1] of size t 
def sample_unique_subsets(n, t, N, seed=0):
    rng = random.Random(int(seed))
    seen = set()
    out = []
    while len(out) < N:
        S = tuple(sorted(rng.sample(range(n), t)))
        if S not in seen:
            seen.add(S)
            out.append(list(S))
    return out
```

Compute the initial solution $\bm{\lambda}_{S,0} = \mathbf{U}_{\mathbf{G}_S}[1,\cdot]$ where $\mathbf{U}_{\mathbf{G}_S}\mathbf{G}_S = \mathbf{R}_{\mathbf{G}_S}$ and $\mathbf{R}_{\mathbf{G}_S}$ is the row HNF of $\mathbf{G}_S$.

```python
def hnf_initial_lambda(G_S):
    """First row of a transformation matrix U with U*G_S = HNF(G_S),
    or None if e_1^T is not in the integer row span of G_S."""
    if G_S.nrows() == 0:
        return None
    R, U = G_S.hermite_form(transformation=True)
    if R.nrows() == 0:
        return None
    e1 = vector(ZZ, [1] + [0]*(G_S.ncols()-1))
    if R.row(0) != e1:
        return None
    return U.row(0)
```

Compute the LLL-reduced kernel basis of $\ker_\mathbb{Z}(\mathbf{G}_S^\top)$.

```python
def kernel_basis_lll_of_GS_transpose(G_S, use_lll=True):
    B = G_S.transpose().right_kernel(ZZ).echelonized_basis_matrix()
    if use_lll and B.nrows() > 1:
        B = B.LLL()
    return B
```

Exact Babai algorithm over $\mathbb{Q}$.

```python
from sage.all import ZZ, QQ, vector

def babai_exact(B, t):
    """Exact-QQ Babai nearest plane. Returns the residual, in t + L(B).

    Differs from babai_nearest: all arithmetic in QQ (no RealField(128) truncation of a lambda_0 that can exceed 1000 bits), 
    and ties round toward +infinity so the residual lands in the half-open fundamental region {sum x_i b*_i : x_i in [-1/2, 1/2)}.
    """
    r, m = B.nrows(), B.ncols()
    if r == 0:
        return vector(ZZ, list(t))
    b  = [vector(QQ, list(B.row(i))) for i in range(r)]
    bs = []
    for i in range(r):
        v = b[i]
        for j in range(i):
            d = bs[j].dot_product(bs[j])
            if d != 0:
                v = v - (b[i].dot_product(bs[j])/d) * bs[j]
        bs.append(v)
    w = vector(QQ, list(t))
    for i in reversed(range(r)):
        d = bs[i].dot_product(bs[i])
        if d == 0:
            continue
        c  = w.dot_product(bs[i]) / d
        ch = (c + QQ(1)/2).floor()
        if ch != 0:
            w = w - QQ(ch)*b[i]
    return vector(ZZ, [ZZ(x) for x in w])

print("babai_exact loaded")
```

Implementing greedy reduction of $\|\bm{\lambda}\|_\infty$ using the kernel basis.

```python
def greedy_reduce_linf(lambda0, B_kernel, passes=3, window=3):
    """
    Greedy reduction of ||lambda||_inf by adding integer kernel vectors.

    Inputs:
      lambda0 : ZZ row vector length s (initial)
      B_kernel: ZZ matrix whose rows b_i satisfy b_i^T * G_S = 0^T
      passes  : number of passes over the basis
      window  : search window around alpha0 (nearest integer)

    Output:
      lambda_best : ZZ row vector
    """
    lam = vector(ZZ, list(lambda0))
    best = linf_norm(lam)

    if B_kernel.nrows() == 0:
        return lam  # no kernel directions available

    for _ in range(int(passes)):
        improved = False

        for i in range(B_kernel.nrows()):
            b = vector(ZZ, list(B_kernel.row(i)))

            # choose index j of largest coordinate of lam
            j = max(range(len(lam)), key=lambda t: abs(lam[t]))

            if b[j] == 0:
                continue  # cannot change the worst coordinate

            # nearest integer to -lam[j]/b[j]
            alpha0 = (QQ(-lam[j])/QQ(b[j]) + QQ(1)/2).floor()

            lam_best = lam
            norm_best = best

            for alpha in range(int(alpha0 - window), int(alpha0 + window) + 1):
                cand = lam + ZZ(alpha) * b
                cand_norm = linf_norm(cand)
                if cand_norm < norm_best:
                    lam_best = cand
                    norm_best = cand_norm

            if norm_best < best:
                lam = lam_best
                best = norm_best
                improved = True

        if not improved:
            break

    return lam
```

Constructing the small $\lambda$.

```python
def construct_small_lambda_babai(G, S, passes=3, window=3):
    """Canonical reconstruction vector, or None if S is unauthorized."""
    S = list(S)
    assert len(S) > 0
    G_S  = G.matrix_from_rows(S)
    lam0 = hnf_initial_lambda(G_S)
    if lam0 is None:
        return None
    lam0 = vector(ZZ, list(lam0))
    B = kernel_basis_lll_of_GS_transpose(G_S, use_lll=True)
    if B.nrows() == 0:
        return lam0
    return greedy_reduce_linf(babai_exact(B, lam0), B, passes=passes, window=window)
```

```python
from itertools import combinations
from sage.all import ZZ

def freq_linf_max_reduced_babai(
    G, s,
    passes=3, window=3,
    return_argmax=False,
    linf_max_freq_size=15,
    wt_max_freq=None,
    verbose_unauth=True
):
    """
    Exact enumeration over all subsets S of size s.

    Subsets S with e_1^T not in the integer row span of G_S are SKIPPED:
    construct_small_lambda_babai returns None for those, and they admit no reconstruction vector at all. 
    The frequency tables therefore describe the distribution conditional on S being authorized, and linf_max is max over authorized S only -- matching the definition of Gamma_1.
    The number skipped is reported separately.

    For lambda_S = construct_small_lambda_babai(G, S, passes, window),
    compute:

      - linf_max = max ||lambda_S||_inf over authorized S
      - ham_weight_at_linf_max = wt_H(lambda_S) for one S attaining linf_max
      - linf_freq = frequency table of ||lambda_S||_inf
      - wt_freq   = frequency table of wt_H(lambda_S)

    Parameters
    ----------
    G : ZZ matrix (n x k)
    s : int
    passes, window : parameters passed to construct_small_lambda_babai
    return_argmax : bool
        If True, also return one subset S achieving linf_max.
    linf_max_freq_size : int or None
        If integer L:
            linf_freq has length L+2
            - indices 0..L count exact values
            - index L+1 is overflow bin for values > L
        If None:
            linf_freq grows dynamically and stores exact frequencies for all values.
    wt_max_freq : int or None
        If None, set to s (full exact table, since wt_H(lambda_S) <= s).
        If integer W:
            wt_freq has length W+2
            - indices 0..W count exact values
            - index W+1 is overflow bin for values > W
    verbose_unauth : bool
        If True, print the number of unauthorized subsets skipped.

    Returns
    -------
    If return_argmax=False:
        (linf_max, ham_weight_at_linf_max, linf_freq, wt_freq)

    If return_argmax=True:
        (linf_max, ham_weight_at_linf_max, linf_freq, wt_freq, S_max)

    The count of skipped subsets is also recorded in the module-level dict
    UNAUTH_LOG, keyed by ('exact', s).
    """
    n = G.nrows()
    s = int(s)
    assert 1 <= s <= n

    # ----- Linf frequency table -----
    if linf_max_freq_size is None:
        linf_freq = []          # dynamic exact table
        linf_overflow = False
    else:
        L = int(linf_max_freq_size)
        linf_freq = [0] * (L + 2)   # 0..L plus overflow at L+1
        linf_overflow = True

    # ----- Hamming-weight frequency table -----
    if wt_max_freq is None:
        # exact table: wt can never exceed s
        W = s
        wt_freq = [0] * (W + 1)     # exact counts for 0..s
        wt_overflow = False
    else:
        W = int(wt_max_freq)
        wt_freq = [0] * (W + 2)     # 0..W plus overflow at W+1
        wt_overflow = True

    linf_max = ZZ(0)
    lam_max = None
    S_max = None
    n_unauth = 0
    n_auth = 0

    for S in combinations(range(n), s):
        lam = construct_small_lambda_babai(G, S, passes=passes, window=window)
        if lam is None:
            n_unauth += 1
            continue
        n_auth += 1
        linf_curr = int(linf_norm(lam))
        wt_curr   = int(hamming_weight(lam))

        # --- update maxima ---
        if lam_max is None or linf_curr > linf_max:
            linf_max = ZZ(linf_curr)
            lam_max = lam
            S_max = list(S)

        # --- update Linf frequency ---
        if linf_overflow:
            if linf_curr <= L:
                linf_freq[linf_curr] += 1
            else:
                linf_freq[L + 1] += 1
        else:
            if linf_curr >= len(linf_freq):
                linf_freq.extend([0] * (linf_curr - len(linf_freq) + 1))
            linf_freq[linf_curr] += 1

        # --- update weight frequency ---
        if wt_overflow:
            if wt_curr <= W:
                wt_freq[wt_curr] += 1
            else:
                wt_freq[W + 1] += 1
        else:
            wt_freq[wt_curr] += 1

    ham_weight_at_linf_max = hamming_weight(lam_max) if lam_max is not None else 0

    UNAUTH_LOG[('exact', s)] = (n_unauth, n_auth)
    if verbose_unauth:
        print("  s=%d [exact]: %d authorized, %d unauthorized skipped"
              % (s, n_auth, n_unauth))

    if return_argmax:
        return (linf_max, ham_weight_at_linf_max, linf_freq, wt_freq, S_max)
    return (linf_max, ham_weight_at_linf_max, linf_freq, wt_freq)
```

Curve wrapper.

```python
def freq_linf_max_reduced_babai_curve(
    G,
    passes=3, window=3,
    s_min=1, s_max=None,
    linf_max_freq_size=15,
    wt_max_freq=None,
    verbose=False,
    verbose_unauth=True
):
    """
    Return a list of tuples:
      [
        (s, linf_max, ham_weight_at_linf_max, linf_freq, wt_freq),
        ...
      ]
    for s in [s_min..s_max].

    Unauthorized subsets (no reconstruction vector) are skipped by freq_linf_max_reduced_babai; 
    their counts are recorded in UNAUTH_LOG under the key ('exact', s), and printed when verbose_unauth is True.
    """
    n = G.nrows()
    if s_max is None:
        s_max = n

    s_min = max(1, int(s_min))
    s_max = min(int(s_max), n)

    out = []
    for s in range(s_min, s_max + 1):
        if verbose:
            print(f"Computing exact frequency table for s={s}")

        result = freq_linf_max_reduced_babai(
            G, s,
            passes=passes,
            window=window,
            return_argmax=False,
            linf_max_freq_size=linf_max_freq_size,
            wt_max_freq=wt_max_freq if wt_max_freq is not None else s,
            verbose_unauth=verbose_unauth
        )
        out.append((s,) + result)

    return out
```

Summary for unauthorized cell.

```python
def unauth_summary(kind=None):
    """Totals from UNAUTH_LOG, aggregated over repeats. kind: 'exact', 'mc', or None."""
    tot = {}
    for k, (nu, na) in UNAUTH_LOG.items():
        if kind is not None and k[0] != kind:
            continue
        s = k[1]
        u, a = tot.get(s, (0, 0))
        tot[s] = (u + nu, a + na)
    print(" s | authorized | unauthorized | frac unauth")
    for s in sorted(tot):
        u, a = tot[s]
        print("%2d | %10d | %12d | %10.4f%%" % (s, a, u, 100.0*u/(a+u) if a+u else 0.0))
    return tot
```

### Monte Carlo simulation

```python
from sage.all import ZZ, binomial

def freq_linf_max_reduced_babai_MC(
    G, s,
    N=1000,
    seed=0,
    passes=3,
    window=3,
    return_argmax=False,
    linf_max_freq_size=15,
    wt_max_freq=None,
    verbose_unauth=True
):
    """
    Monte Carlo version of frequency tables for reduced reconstruction vectors using the Babai + greedy approach.

    Subsets S with e_1^T not in the integer row span of G_S are SKIPPED: construct_small_lambda_babai returns None for those, and they admit no reconstruction vector at all. 
    The frequency tables therefore describe the distribution conditional on S being authorized, and observed_linf_max is a max over authorized S only
    -- matching the definition of Gamma_1. 
    The number skipped is recorded in UNAUTH_LOG under the key ('mc', s, seed).

    Note the tables no longer sum to N: they sum to the number of authorized subsets in the sample.

    Over N distinct sampled subsets S of size s, where
        lambda_S = construct_small_lambda_babai(G, S, passes, window),
    compute:

      - observed_linf_max = max ||lambda_S||_inf over the sample
      - ham_weight_at_linf_max = wt_H(lambda_S) for one sampled maximizer
      - linf_freq = sample frequency table of ||lambda_S||_inf
      - wt_freq   = sample frequency table of wt_H(lambda_S)

    Parameters
    ----------
    G : ZZ matrix (n x k)
    s : int
    N : int
        Number of distinct sampled subsets of size s
    seed : int
    passes, window : parameters passed to construct_small_lambda_babai
    return_argmax : bool
        If True, also return one sampled subset S attaining observed_linf_max
    linf_max_freq_size : int or None
        If integer L:
          linf_freq has length L+2:
            indices 0..L count exact values
            index L+1 is overflow bin for values > L
        If None:
          linf_freq grows dynamically and stores all sampled values exactly.
    wt_max_freq : int or None
        If None, defaults to s.
        If integer W:
          wt_freq has length W+2:
            indices 0..W count exact values
            index W+1 is overflow bin for values > W
    verbose_unauth : bool
        If True, print the number of unauthorized subsets skipped.

    Returns
    -------
    If return_argmax=False:
        (observed_linf_max, ham_weight_at_linf_max, linf_freq, wt_freq)

    If return_argmax=True:
        (observed_linf_max, ham_weight_at_linf_max, linf_freq, wt_freq, S_max)
    """
    n = G.nrows()
    s = int(s)
    assert 1 <= s <= n

    total = int(binomial(n, s))
    if N > total:
        raise ValueError(f"N={N} exceeds number of distinct subsets C({n},{s})={total}")

    # ----- Linf frequency table -----
    if linf_max_freq_size is None:
        linf_freq = []          # dynamic exact table
        linf_overflow = False
    else:
        L = int(linf_max_freq_size)
        linf_freq = [0] * (L + 2)   # 0..L plus overflow at L+1
        linf_overflow = True

    # ----- Hamming-weight frequency table -----
    if wt_max_freq is None:
        W = s
        wt_freq = [0] * (W + 1)     # exact table for 0..s
        wt_overflow = False
    else:
        W = int(wt_max_freq)
        wt_freq = [0] * (W + 2)     # 0..W plus overflow at W+1
        wt_overflow = True

    observed_linf_max = ZZ(0)
    lam_max = None
    S_max = None
    n_unauth = 0
    n_auth = 0

    samples = sample_unique_subsets(n, s, N, seed)

    for S in samples:
        lam = construct_small_lambda_babai(G, S, passes=passes, window=window)
        if lam is None:
            n_unauth += 1
            continue
        n_auth += 1
        linf_curr = int(linf_norm(lam))
        wt_curr   = int(hamming_weight(lam))

        # ---- update observed maximum ----
        if lam_max is None or linf_curr > observed_linf_max:
            observed_linf_max = ZZ(linf_curr)
            lam_max = lam
            S_max = list(S)

        # ---- update Linf frequency ----
        if linf_overflow:
            if linf_curr <= L:
                linf_freq[linf_curr] += 1
            else:
                linf_freq[L + 1] += 1
        else:
            if linf_curr >= len(linf_freq):
                linf_freq.extend([0] * (linf_curr - len(linf_freq) + 1))
            linf_freq[linf_curr] += 1

        # ---- update weight frequency ----
        if wt_overflow:
            if wt_curr <= W:
                wt_freq[wt_curr] += 1
            else:
                wt_freq[W + 1] += 1
        else:
            wt_freq[wt_curr] += 1

    ham_weight_at_linf_max = hamming_weight(lam_max) if lam_max is not None else 0

    UNAUTH_LOG[('mc', s, int(seed))] = (n_unauth, n_auth)
    if verbose_unauth:
        print("  s=%d [mc, seed=%d]: %d authorized, %d unauthorized skipped"
              % (s, int(seed), n_auth, n_unauth))

    if return_argmax:
        return (observed_linf_max, ham_weight_at_linf_max, linf_freq, wt_freq, S_max)
    return (observed_linf_max, ham_weight_at_linf_max, linf_freq, wt_freq)
```

Curve wrapper.

```python
from sage.all import ZZ

def freq_linf_max_reduced_babai_curve_MC(
    G,
    N=1000,
    seed=0,
    repeats=3,
    passes=3,
    window=3,
    s_min=1,
    s_max=None,
    linf_max_freq_size=15,
    wt_max_freq=None,
    verbose=False,
    verbose_unauth=True
):
    """
    Monte Carlo curve wrapper for frequency tables using the canonical exact-Babai + greedy reconstruction pipeline.

    For each s in [s_min..s_max], run
        freq_linf_max_reduced_babai_MC(...)
    'repeats' times with seeds seed + 1000*s + r, and aggregate the sample
    frequency tables.

    Unauthorized subsets are skipped by freq_linf_max_reduced_babai_MC, so the aggregated tables sum to the number of AUTHORIZED subsets in the sample, not to N*repeats. 
    Per-repeat counts are recorded in UNAUTH_LOG under keys ('mc', s, seed_sr); use unauth_summary('mc') to total them.

    Returns
    -------
    A list of tuples:
      [
        (s, observed_linf_max, ham_weight_at_linf_max, linf_freq, wt_freq),
        ...
      ]

    where:
      - observed_linf_max is the maximum observed over all repeats
      - ham_weight_at_linf_max is the weight of one sampled maximizer
      - linf_freq is the aggregated sample frequency table across repeats
      - wt_freq is the aggregated sample frequency table across repeats
    """
    n = G.nrows()
    if s_max is None:
        s_max = n

    s_min = max(1, int(s_min))
    s_max = min(int(s_max), n)

    curve = []

    for s in range(s_min, s_max + 1):
        best_linf = ZZ(0)
        best_wt = 0

        agg_linf_freq = None
        agg_wt_freq = None

        for r in range(repeats):
            seed_sr = int(seed) + 1000*s + r

            linf_s, wt_s, linf_freq_s, wt_freq_s = freq_linf_max_reduced_babai_MC(
                G, s,
                N=N,
                seed=seed_sr,
                passes=passes,
                window=window,
                return_argmax=False,
                linf_max_freq_size=linf_max_freq_size,
                wt_max_freq=wt_max_freq if wt_max_freq is not None else s,
                verbose_unauth=verbose_unauth
            )

            # ---- update observed max over repeats ----
            if linf_s > best_linf:
                best_linf = linf_s
                best_wt = wt_s

            # ---- aggregate Linf frequency tables ----
            if agg_linf_freq is None:
                agg_linf_freq = list(linf_freq_s)
            else:
                if len(linf_freq_s) > len(agg_linf_freq):
                    agg_linf_freq.extend([0] * (len(linf_freq_s) - len(agg_linf_freq)))
                elif len(agg_linf_freq) > len(linf_freq_s):
                    linf_freq_s = list(linf_freq_s) + [0] * (len(agg_linf_freq) - len(linf_freq_s))
                agg_linf_freq = [a + b for a, b in zip(agg_linf_freq, linf_freq_s)]

            # ---- aggregate weight frequency tables ----
            if agg_wt_freq is None:
                agg_wt_freq = list(wt_freq_s)
            else:
                if len(wt_freq_s) > len(agg_wt_freq):
                    agg_wt_freq.extend([0] * (len(wt_freq_s) - len(agg_wt_freq)))
                elif len(agg_wt_freq) > len(wt_freq_s):
                    wt_freq_s = list(wt_freq_s) + [0] * (len(agg_wt_freq) - len(wt_freq_s))
                agg_wt_freq = [a + b for a, b in zip(agg_wt_freq, wt_freq_s)]

        if verbose:
            print(f"s={s}, observed max over repeats = {best_linf}")

        curve.append((s, best_linf, best_wt, agg_linf_freq, agg_wt_freq))

    return curve
```

# Data visualization

Grouped bar chart (grouped histogram).

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_grouped_bar_distribution(curve_data, which="linf", selected_s=None,
                                  normalize=True, max_x=None,
                                  overflow_label=True, figsize=(10, 5)):
    # ---- filter s values ----
    if selected_s is not None:
        curve_data = [row for row in curve_data if row[0] in selected_s]
    if not curve_data:
        raise ValueError("No matching s values found in curve_data")

    # ---- choose frequency table ----
    freq_idx = 3 if which == "linf" else 4
    xlabel = r"$\|\lambda_S\|_\infty$" if which == "linf" else r"$wt_H(\lambda_S)$"
    s_values = [row[0] for row in curve_data]
    freq_lists = [list(row[freq_idx]) for row in curve_data]

    # ---- determine x-range ----
    max_len = max(len(f) for f in freq_lists)
    if max_x is None:
        max_x = max_len - 1
    else:
        max_x = min(int(max_x), max_len - 1)

    # ---- build matrix ----
    M = []
    for f in freq_lists:
        row = f + [0] * (max_len - len(f))
        row = row[:max_x + 1]
        if normalize:
            total = sum(f)
            if total > 0:
                row = [x / total for x in row]
        M.append(row)
    M = np.array(M)

    # ---- BW-friendly styles ----
    # Evenly spaced grays from dark to light (avoid pure white — bars vanish)
    n_series = len(s_values)
    gray_levels = np.linspace(0.15, 0.75, n_series)
    bw_colors = [(g, g, g) for g in gray_levels]

    # Distinct hatch patterns — readable even on low-contrast printers
    hatch_patterns = ['', '///', '...', 'xxx', '\\\\\\', '+++', 'ooo', '---']

    # ---- grouped bars ----
    n_groups = max_x + 1
    x = np.arange(n_groups)
    width = 0.8 / n_series

    fig, ax = plt.subplots(figsize=figsize)
    for i, s in enumerate(s_values):
        ax.bar(
            x + i * width,
            M[i],
            width=width,
            label=f"$s={s}$",
            color=bw_colors[i % len(bw_colors)],
            hatch=hatch_patterns[i % len(hatch_patterns)],
            edgecolor='black',
            linewidth=0.6,
        )

    # ---- x tick labels ----
    labels = [str(i) for i in range(n_groups)]
    if overflow_label and which == "linf" and n_groups >= 2:
        labels[-1] = f">${n_groups - 2}$"

    ax.set_xticks(x + width * (n_series - 1) / 2)
    ax.set_xticklabels(labels, rotation=45)
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Relative frequency" if normalize else "Count")
    ax.set_title(f"Distribution of {xlabel} across multiple $s$")
    ax.legend(framealpha=1, edgecolor='black')
    ax.set_facecolor('white')
    fig.patch.set_facecolor('white')

    plt.tight_layout()
    plt.show()
```

```python
def plot_linf_bar_for_s(curve_data, s_target, normalize=True, overflow_label=True, figsize=(8, 4)):
    """
    Plot bar chart of linf frequency distribution for one chosen s.
    """
    entry = None
    for row in curve_data:
        if row[0] == s_target:
            entry = row
            break
    if entry is None:
        raise ValueError(f"s={s_target} not found in curve_data")

    freq = list(entry[3])
    total = sum(freq)
    vals = [x / total for x in freq] if normalize and total > 0 else freq

    labels = [str(i) for i in range(len(freq))]
    if overflow_label and len(freq) >= 2:
        labels[-1] = f">{len(freq)-2}"

    plt.figure(figsize=figsize)
    plt.bar(range(len(vals)), vals, width=0.8)
    plt.xticks(range(len(vals)), labels, rotation=45)
    plt.xlabel(r"$\|\lambda_S\|_\infty$")
    plt.ylabel("Relative frequency" if normalize else "Count")
    plt.title(f"Infinity norm distribution for s={s_target}")
    plt.tight_layout()
    plt.show()
```
